```python
video_path = "/kaggle/input/datasets/nazaninashrafi/video-samples/15186864_1080_1920_25fps.mp4"
cap = cv.VideoCapture(video_path)

lower_orange = np.array([10, 100, 100])
upper_orange = np.array([20, 255, 255])


while True:
    ret, frame = cap.read()
    if not ret:
        print("faile to load")
        break

    hsv = cv.cvtColor(frame, cv.COLOR_BGR2HSV)

    orange_mask = cv.inRange(hsv, lower_orange, upper_orange)

    contours, _ = cv.findContours(
        orange_mask,
        cv.RETR_EXTERNAL,
        cv.CHAIN_APPROX_SIMPLE
    )

    for cnt in contours:
        if cv.contourArea(cnt) < 800:
            continue

        x,y,w,h = cv.boundingRect(cnt)
        cv.rectangle(
            frame,
            (x,y),
            (x + w, y + h),
            (0,255,0),
            5
        )

    rgb_frame = cv.cvtColor(frame, cv.COLOR_BGR2RGB)
    plt.imshow(rgb_frame)
    plt.axis("off")

    break

cap.release()
plt.show()
```

```python
video_path = "/kaggle/input/datasets/nazaninashrafi/video-samples/15186864_1080_1920_25fps.mp4"
cap = cv.VideoCapture(video_path)

output_frames = []

lower_orange = np.array([10, 100, 100])
upper_orange = np.array([20, 255, 255])

while len(output_frames) < 6:
    ret, frame = cap.read()
    if not ret:
        print("faile to load")
        break


    hsv = cv.cvtColor(frame, cv.COLOR_BGR2HSV)

    orange_mask = cv.inRange(hsv, lower_orange, upper_orange)

    contours, _ = cv.findContours(
        orange_mask,
        cv.RETR_EXTERNAL,
        cv.CHAIN_APPROX_SIMPLE
    )

    for cnt in contours:
        if cv.contourArea(cnt) < 800:
            continue

        x,y,w,h = cv.boundingRect(cnt)
        cv.rectangle(
            frame,
            (x,y),
            (x + w, y + h),
            (0,255,0),
            5
        )

    rgb_frame = cv.cvtColor(frame, cv.COLOR_BGR2RGB)

    output_frames.append(rgb_frame)


cap.release()

plt.figure(figsize=(10,10))
for i, img in enumerate(output_frames):
    plt.subplot(2,3, i+1)
    plt.imshow(img)
    plt.title(f"Detected Motion - Frame {i}")
    plt.axis('off')
plt.show()
```

```python
video_path = "/kaggle/input/datasets/shawon10/road-traffic-video-monitoring/dhaka_traffic.mp4"
cap = cv.VideoCapture(video_path)

lower_red1 = np.array([0,90,170])
upper_red1 = np.array([10,255,255])

lower_red2 = np.array([170,70,50])
upper_red2 = np.array([179,255,255])

kernel = np.ones((5,5), np.uint8)

output_frames = []

while len(output_frames) < 10:
    ret, frame = cap.read()
    if not ret:
        print("faile to load")
        break

    hsv = cv.cvtColor(frame, cv.COLOR_BGR2HSV)

    red_mask1 = cv.inRange(hsv, lower_red1, upper_red1)
    red_mask2 = cv.inRange(hsv, lower_red2, upper_red2)
    red_mask = red_mask1 | red_mask2
    red_mask = cv.morphologyEx(red_mask, cv.MORPH_OPEN, kernel)
    red_mask = cv.morphologyEx(red_mask, cv.MORPH_CLOSE, kernel)

    

    # rgb_frame = cv.cvtColor(frame, cv.COLOR_BGR2RGB)
    # result = cv.bitwise_and(rgb_frame, rgb_frame, mask=red_mask)
    # plt.imshow(result)
    # plt.axis("off")
    # plt.show()

    contours, _ = cv.findContours(
        red_mask,
        cv.RETR_EXTERNAL,
        cv.CHAIN_APPROX_SIMPLE
    )


    # rgb_frame = cv.cvtColor(frame, cv.COLOR_BGR2RGB)
    # plt.imshow(rgb_frame)
    # plt.axis("off")
    # plt.show()

    for cnt in contours:
        if cv.contourArea(cnt) < 200:
            continue
    
        x,y,w,h = cv.boundingRect(cnt)
        cv.rectangle(
            frame,
            (x,y),
            (x+w, y+h),
            (0,255,0),
            2
        )

    rgb_frame = cv.cvtColor(frame, cv.COLOR_BGR2RGB)
    # plt.imshow(rgb_frame)
    # plt.axis("off")
    # plt.show()

    output_frames.append(rgb_frame)



    # break

cap.release()

plt.figure(figsize=(10,20))
for i, img in enumerate(output_frames):
    plt.subplot(5, 2, i+1)
    plt.imshow(img)
    plt.title(f"Detected Motion - Frame {i}")
    plt.axis('off')
plt.show()
```
